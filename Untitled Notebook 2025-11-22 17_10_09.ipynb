```sql

-- Create dimension-ready views in Silver layer


--DIM_DATE view
    CREATE OR REPLACE VIEW dim_date AS
    SELECT DISTINCT
        date_key,
        date_occurred_clean as full_date,
        day_of_week,
        day_name,
        day_of_month,
        week_of_year,
        month,
        month_name,
        quarter,
        year,
        is_weekend
    FROM workspace.bi_demo.la_crime_clean
    WHERE date_occurred_clean IS NOT NULL;


--DIM_TIME view
    CREATE OR REPLACE VIEW dim_time AS
    SELECT DISTINCT
        time_key,
        hour_occurred as hour,
        minute_occurred as minute,
        time_period,
        CONCAT(
            LPAD(hour_occurred, 2, '0'), ':',
            LPAD(minute_occurred, 2, '0'), '-',
            LPAD(hour_occurred, 2, '0'), ':59'
        ) as time_bucket
    FROM workspace.bi_demo.la_crime_clean
    WHERE time_key IS NOT NULL;


--DIM_CRIME view

    CREATE OR REPLACE VIEW dim_crime AS
    SELECT DISTINCT
        crime_code_clean as crm_cd,
        crime_desc as crm_cd_desc,
        crime_part_clean as part_1_2,
        crime_category,
        mocodes
    FROM workspace.bi_demo.la_crime_clean
    WHERE crime_code_clean IS NOT NULL;


--DIM_LOCATION view
    CREATE OR REPLACE VIEW dim_location AS
    SELECT DISTINCT
        area_clean as area,
        area_name,
        rpt_dist_no_clean as rpt_dist_no,
        premis_code_clean as premis_cd,
        premis_desc,
        location_address_clean as location_address,
        cross_street,
        latitude_clean as lat,
        longitude_clean as lon,
        has_valid_coordinates as crime_hotspot_flag,
        CASE 
            WHEN data_quality_score >= 3 THEN 'High Priority'
            WHEN data_quality_score >= 2 THEN 'Medium Priority'
            ELSE 'Low Priority'
        END as crime_severity_zone
    FROM workspace.bi_demo.la_crime_clean;
--DIM_VICTIM view
    CREATE OR REPLACE VIEW dim_victim AS
    SELECT DISTINCT
        victim_age_clean as vict_age,
        age_group,
        victim_sex_clean as vict_sex,
        victim_descent as vict_descent,
        descent_description
    FROM workspace.bi_demo.la_crime_clean
    WHERE has_victim_info = 'Y';

--DIM_STATUS view
    CREATE OR REPLACE VIEW dim_status AS
    SELECT DISTINCT
        status_clean as status,
        status_desc,
        arrest_flag as is_arrest,
        COALESCE(arrest_type, 'No Arrest') as case_resolution_type
    FROM workspace.bi_demo.la_crime_clean;

--DIM_WEAPON view
    CREATE OR REPLACE VIEW dim_weapon AS
    SELECT DISTINCT
        weapon_code_clean as weapon_used_cd,
        weapon_desc,
        weapon_category
    FROM workspace.bi_demo.la_crime_clean
    WHERE weapon_code_clean IS NOT NULL;



--Generate SQL for creating tables in Snowflake


SELECT * FROM dim_date;



SELECT * FROM dim_time;

SELECT 
    ROW_NUMBER() OVER (ORDER BY crm_cd) as crime_key,
    * 
FROM dim_crime;


SELECT 
    ROW_NUMBER() OVER (ORDER BY area, premis_cd) as location_key,
    * 
FROM dim_location;



SELECT 
    ROW_NUMBER() OVER (ORDER BY vict_age, vict_sex, vict_descent) as victim_key,
    * 
FROM dim_victim;

SELECT 
    ROW_NUMBER() OVER (ORDER BY status) as status_key,
    * 
FROM dim_status;



SELECT 
    ROW_NUMBER() OVER (ORDER BY weapon_used_cd) as weapon_key,
    * 
FROM dim_weapon;



SELECT 
    dr_no_clean as dr_no,
    date_key,
    time_key,
    -- Location lookup will be done in Snowflake
    area_clean as area_lookup,
    premis_code_clean as premis_lookup,
    -- Victim lookup
    victim_age_clean as age_lookup,
    victim_sex_clean as sex_lookup,
    victim_descent as descent_lookup,
    -- Crime lookup
    crime_code_clean as crime_lookup,
    -- Weapon lookup
    weapon_code_clean as weapon_lookup,
    -- Status lookup
    status_clean as status_lookup,
    -- Measures
    date_reported_clean as date_reported,
    arrest_flag,
    arrest_type
FROM workspace.bi_demo.la_crime_clean;

```
